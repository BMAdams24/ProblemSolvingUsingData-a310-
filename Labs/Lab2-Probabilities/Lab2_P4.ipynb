# **Lab 2 Probabilities**

**4.** So far, we've only analyzed event conditions that satisfy some single value. Now we'll analyze event conditions that span intervals of values. Write code that will compute the likelihood that our eight consecutive die rolls sum to a value between 14 and 24 inclusive.

```python
from itertools import product
from collections import defaultdict

def get_matching_event(condition, outcomes):
    return {value for value in outcomes if condition(value)}

def compute_event_probability(condition, outcomes):
    matching = get_matching_event(condition, outcomes)

    if isinstance(outcomes, set):
        return len(matching) / len(outcomes)

    return (
        sum(outcomes[value] for value in matching)
        / sum(outcomes.values())
    )

dice_values = list(range(1, 7))
print(dice_values)

all_rolls = set(product(dice_values, repeat=8))

weighted_sample_space = defaultdict(int)

for rolls in all_rolls:
    weighted_sample_space[sum(rolls)] += 1

def in_interval(value, low, high):
    return low <= value <= high

probability = compute_event_probability(
    lambda total: in_interval(total, 14, 24),
    weighted_sample_space
)

print(f"Probability of interval is {probability}")
print(f"Percentage: {probability * 100:.2f}%")
```
