# **Lab 2 Probabilities**

**1.** Certain actions have measurable outcomes. A sample space is the set of all possible outcomes an action could produce. Let's take the simple action of flipping a coin. The coin will land on either heads or tails. An event is the subset of those elements within a sample space that satisfy some event condition. Follow the presentation in the book and (a) define a sample space for a fair, unbiased coin; (b) define a weighted sample space for a biased coin that generates tails three times more often than heads; and (c) write a function compute_event_probability that calculates the probability of each elementary event (heads, tails, either, or neither) in both cases.

```python
sample_space = {'Heads', 'Tails'}

def is_heads_or_tails(result):
    return result in sample_space

def is_neither(result):
    return result not in sample_space

def is_heads(result):
    return result == "Heads"

def is_tails(result):
    return result == "Tails"

def get_matching_event(condition, outcomes):
    return {item for item in outcomes if condition(item)}

event_conditions = [is_heads_or_tails, is_heads, is_tails, is_neither]

weighted_sample_space = {
    'Heads': 1,
    'Tails': 3
}

def compute_event_probability(condition, outcomes):
    matches = get_matching_event(condition, outcomes)

    if isinstance(outcomes, set):
        return len(matches) / len(outcomes)

    match_weight = sum(outcomes[item] for item in matches)
    total_weight = sum(outcomes.values())

    return match_weight / total_weight

print("Biased coin results (tails is 3 times more likely than heads):")
for condition in event_conditions:
    probability = compute_event_probability(condition, weighted_sample_space)
    print(f"{condition.__name__}: probability = {probability} ({probability * 100:.2f}%)")

print("\nFair coin results (heads and tails are equally likely):")
for condition in event_conditions:
    probability = compute_event_probability(condition, sample_space)
    print(f"{condition.__name__}: probability = {probability} ({probability * 100:.2f}%)")
```
