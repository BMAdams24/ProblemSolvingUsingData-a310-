# **Lab 2 Probabilities**

**2.** We will now solve several example problems using compute_event_probability. We start by assuming that a family has four children. What is the probability that exactly two of the children are girls? We'll assume that each child is equally likely to be either a boy or a girl. (You can reformulate the problem as follows: you have four coins, each is equaly likely to be heads or tails, what is the chance the set is balanced (i.e., two heads and two tails)?).

```python
from itertools import product

def get_matching_event(condition, outcomes):
    return {item for item in outcomes if condition(item)}

def compute_event_probability(condition, outcomes):
    matches = get_matching_event(condition, outcomes)

    if isinstance(outcomes, set):
        return len(matches) / len(outcomes)

    return (
        sum(outcomes[item] for item in matches)
        / sum(outcomes.values())
    )

children = ["Boy", "Girl"]

sample_space = set(product(children, repeat=4))

def has_two_girls(family):
    return sum(child == "Girl" for child in family) == 2

probability = compute_event_probability(has_two_girls, sample_space)

print(f"Probability of 2 girls is {probability}")
print(f"Percentage: {probability * 100:.2f}%")
```
