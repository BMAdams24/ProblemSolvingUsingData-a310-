**3.** Suppose we're shown a fair six-sided die whose faces are numbered from 1 to 6. The die is rolled eight times. What is the probability that these eight die rolls add up to 28?

```python
from itertools import product

def get_matching_event(condition, outcomes):
    return {item for item in outcomes if condition(item)}

def compute_event_probability(condition, outcomes):
    event = get_matching_event(condition, outcomes)

    if isinstance(outcomes, set):
        return len(event) / len(outcomes)

    return (
        sum(outcomes[item] for item in event)
        / sum(outcomes.values())
    )

rolls = [1, 2, 3, 4, 5, 6]
print(rolls)

sample_space = set(product(rolls, repeat=8))
print(len(sample_space))

def sum_is_28(roll_result):
    return sum(roll_result) == 28

probability = compute_event_probability(sum_is_28, sample_space)

print(f"8 rolls sum to 28 with a probability of {probability}")
print(f"Percentage: {probability * 100:.2f}%")

check = compute_event_probability(
    lambda rolls: sum(rolls) == 28,
    sample_space
)

assert probability == check
```
