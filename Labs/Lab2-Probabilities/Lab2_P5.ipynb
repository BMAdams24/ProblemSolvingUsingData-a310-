# **Lab 2 probabilities**

**5.** Evaluating extremes using interval analysis. What is the probability that 10 fair coin flips produce either 0 to 2 heads or 8 to 10 heads? (Alternatively stated, you need to write code that finds out the probability that the coin flips do not produce from 3 to 7 heads.) Now determine the probability of 20 fair coin flips not producing from 5 to 15 heads (or tails). What does the book say about Chapter 2?

```python
from itertools import product
from collections import defaultdict

def get_matching_event(condition, outcomes):
    return {item for item in outcomes if condition(item)}

def compute_event_probability(condition, outcomes):
    matching = get_matching_event(condition, outcomes)

    if isinstance(outcomes, set):
        return len(matching) / len(outcomes)

    return (
        sum(outcomes[item] for item in matching)
        / sum(outcomes.values())
    )

def generate_coin_sample_space(num_flips=10):
    counts = defaultdict(int)

    for sequence in product(["Heads", "Tails"], repeat=num_flips):
        heads = sum(side == "Heads" for side in sequence)
        counts[heads] += 1

    return counts

weighted_sample_space = generate_coin_sample_space()

assert weighted_sample_space[10] == 1
assert weighted_sample_space[9] == 10

def in_interval(value, minimum, maximum):
    return minimum <= value <= maximum

probability = compute_event_probability(
    lambda heads: in_interval(heads, 8, 10),
    weighted_sample_space
)

print(f"Probability of observing more than 7 heads is {probability}")
print(f"Percentage: {probability * 100:.2f}%")

probability = compute_event_probability(
    lambda heads: not in_interval(heads, 3, 7),
    weighted_sample_space
)

print(f"Probability of observing more than 7 heads or 7 tails is {probability}")
print(f"Percentage: {probability * 100:.2f}%")

weighted_sample_space_20 = generate_coin_sample_space(20)

probability = compute_event_probability(
    lambda heads: not in_interval(heads, 5, 15),
    weighted_sample_space_20
)

print(f"Probability of observing more than 15 heads or 15 tails is {probability}")
print(f"Percentage: {probability * 100:.2f}%")

print(
    "Chapter 2 uses Matplotlib to plot probability distributions "
    "and compare coin-flip probabilities."
)
```
