```python
# 24. solve exercise 3-1 on page 57

# write your solution here
## Dependencies
import numpy as np
## Data: yearly salary in ($1000) [2017, 2018, 2019]
alice = [99, 101, 103]
bob = [110, 108, 105]
tim = [90, 88, 85]
names = ['alice', 'bob', 'tim']
salaries = np.array([alice, bob, tim])
taxation = np.array([[0.2, 0.25, 0.22],
[0.4, 0.5, 0.5],
[0.1, 0.2, 0.1]])
## One-liner
max_income = np.max(salaries - salaries * taxation)

#for lab
after_tax_income = (salaries - salaries * taxation).max(axis=1)

max_index = np.argmax(after_tax_income)
max_name = names[max_index]

## Result
print(max_income)

print("The person with the max income was: ", max_name)
```
