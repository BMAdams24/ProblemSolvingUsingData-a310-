### **Pig Latin Rules:**

- If a word begins with a vowel, add `way` to the end.
- If a word begins with a consonant followed by a vowel, move the first consonant to the end and add `ay`.
- If a word begins with two consonants, move the first two consonants to the end and add `ay`.
- Non-letter characters at the end of a word should remain at the end.
- The original capitalization of letters should be preserved.

```python
import re

def pig_latin_word(word):
    consonant = r'[BCDFGHJKLMNPQRSTVWXYZbcdfghjklmnpqrstvwxyz]'
    vowel = r'[AEIOUaeiou]'

    # Case1
    match = re.fullmatch(
        rf'({consonant})({vowel})(\w*)(\W*)',
        word
    )

    if match:
        return (
            match.group(2)
            + match.group(3)
            + match.group(1)
            + "ay"
            + match.group(4)
        )
    #Case2
    match = re.fullmatch(
        rf'({consonant})({consonant})(\w*)(\W*)',
        word
    )

    if match:
        return (
            match.group(3)
            + match.group(1)
            + match.group(2)
            + "ay"
            + match.group(4)
        )
    #Case3
    match = re.fullmatch(
        rf'({vowel})(\w*)(\W*)',
        word
    )

    if match:
        return (
            match.group(1)
            + match.group(2)
            + "way"
            + match.group(3)
        )

    return word
```
