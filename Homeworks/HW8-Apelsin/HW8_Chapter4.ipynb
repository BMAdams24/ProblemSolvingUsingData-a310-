**8. What is mathematical induction? IS it being used here?**

Mathematical induction is a proof technique that shows if something works for a small case, shows it working for the next case, then concluding it works for all cases. The chapter uses inductive reasoning by showing 2-card deck fails, 4-card deck fails, --> 10 card deck fails, so its reasonable to expect 52-card deck to also fail.
