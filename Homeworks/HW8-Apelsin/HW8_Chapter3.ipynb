**1. What is chapter 3 about? What is the main goal?**

This chapter introduces random simulation using NumPy and instead of calculating probabilities through sample spaces, it's estimating them by simulating thousands of random events. This is usefull when exact calculations end up being too difficult or hard on the computer.

**2. Why can we now read it and make sense of it?**

Since the previous chapters introduced sample spaces, event probability &distributions, and histogtams, this chapter expands on those ideas by showing how to esitmate probabilities through simulation instead of counting.

**3. Why do we learn to simulate random coin flips and die rolls?**

Since coin flips and die rolls are simple random occurances, we use them to provide a foundation before moving on to more complicated scenarios like card games. The randomness and repeated trials are good simulation examples.

**4. Why are we interested in biased coin flips?**

We are interested in biased coin flips due to biased coins modeling simulation outcomes that are not equally likely. An example would be a biased coin that has a 70% for heads instead of a 50%.

**5. What is a confidence interval and why is it important?**

A confidence interval is a range of values that most likely contain the true probability and tells us how certain we are about the estimate. If we dont use confidence intervals, then all we have is a single estimate with no measure of its reliability.

**6. Why do we draw histograms? What do we get from them?**

Histograms group similar values into bins and show the distribution of observations. This reveals where the values are concentrated and help compute confidence intervals, making it easy to see.

**7. What is the central limit theorem and what does it say?**

The central limit theorem states that "If we repetedly take many samples from a population, the distribution of sample mean tends to become a normal distribution". It also predicts larger sample sizes and the estimates become more precise as sampling increases.

**8. Can you compute a histogram's parameters without drawing a plot?**

Yes, using np.histogram() we can return bin edges and likelihoods/counts without creating a graph

**9. How is all this relevant to our card game?**

The original card game is too complex to solve only using brute force sample spaces. So, this simulation allows the ability to shuffle many decks, test strategies, estimate win probabilities, and measure confidence in those estimates.

**10. How does this chapter propose we shuffle our deck of cards?**

Using NumPy we can either use np.random.shuffle(deck) or np.random.permutation(deck) since a permutation is simply a random arrangement of cards.

**11. Why are we all of a sudden working with a 10-card deck?**

Since a normal 52-card deck has too many possible arrangements, using a 10-card deck has only 252 unique color arrangements, can be completely analyzed, and still behaves similarly. Essentially, it makes exact probabilty calculations possible.
