**Problem 1**. A fair game has an expected value of 0. When rolling two dice, there are 36 possible outcomes and 6 of them add up to 7, so the probability of winning is 6/36 = 1/6. The probability of losing is 5/6. The expected value is:

4(1/6) + (-1)(5/6) = (4/6)-(5/6) = -(1/6)

Since the expected value is negative, the game is not fair for the player. The house has the advantage. To make the game fair, the winning payout would need to be 5 chips instead of 4.

```python
import random

times = 1000000
money = 0

for i in range(times):
  d1 = random.randint(1, 6)
  d2 = random.randint(1, 6)
  sum = d1 + d2

  if sum == 7:
    money += 4
  else:
    money -= 1

print(money/times)
```

**Problem 2**

The program calculates the probability that the sum of three dice is not a prime number. The three nested loops generate every possible outcome when rolling three dice. Since each die has 6 possible values, there are

6(6)(6)=2166 total outcomes.

For each outcome, the code computes the sum of the three dice. If the sum is not one of the possible prime sums

3,5,7,11,13,17

the counter is increased by one. After checking all 216 outcomes, the program divides the number of favorable outcomes by 216 to obtain the probability.

The output is approximately

0.662037

This means there is about a 66.2% chance that the sum of three dice is not a prime number.

```python
count = 0

for d1 in range(1, 7):
  for d2 in range(1, 7):
    for d3 in range(1, 7):
      sum = d1 + d2 + d3

      if sum not in [3, 5, 7, 11, 13, 17]:
        count += 1
      else:
        pass

print(count/(6 * 6 * 6))
```

**Problem 3** Each locker is toggled once for every divisor that locker number has.

Most numbers have divisors that come in pairs. For example:

12=1(12)=2(6)=3(4)

Since 12 has an even number of divisors, it gets toggled an even number of times and ends up closed.

Perfect squares are different because one divisor pair repeats. For example:

36 = 1(36) --> 2(18) --> 3(12) --> 4(9) --> 6(6)

The divisor 6 is counted only once, giving 36 an odd number of divisors. Therefore, perfect squares are toggled an odd number of times and remain open.

The lockers left open after all 100 students have passed are:

1,4,9,16,25,36,49,64,81,100

These are exactly the perfect squares from 1 to 100.

**Extra code for problem 3 just for fun:**

```python
lockers = [False] * 100

for student in range(1, 101):

    for locker in range(student - 1, 100, student):
        lockers[locker] = not lockers[locker]

    open_lockers = []

    for i in range(100):
        if lockers[i]:
            open_lockers.append(i + 1)

    print("After student", student, ":", open_lockers)

print("\nFinal open lockers:")

for i in range(100):
    if lockers[i]:
        print(i + 1, end=" ")
```
