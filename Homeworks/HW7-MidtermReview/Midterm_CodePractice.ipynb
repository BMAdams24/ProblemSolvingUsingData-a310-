```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class LinkedList:

  def __init__(self):
    self.head = None


  def insert(self, value):

    new = Node(value)

    if self.head == None:
      self.head = new
      return

    current = self.head

    while current.next:
      current = current.next

    current.next = new


  def remove(self, value):
    if self.head.data == value:
      self.head = self.head.next
      return

    current = self.head

    while current.next:
      if current.next.data == value:
        current.next = current.next.next
        return

      current = current.next


  def find(self, value):
    current = self.head

    while current:
      if current.data == value:
        return True

      current = current.next

    return False


  def show(self):
    current = self.head

    while current:
      print(current.data)
      current = current.next
```

# **Sort(longest, rest, merge)**

```python
def longest(num):

  if len(num) <= 1: #if the length is less than 1, return the seq (only 1 number)
    return num

  if num[0] <= num[1]: #if first element is less than the next
    return [num[0]] + longest(num[1:]) #return first element and run the longest again for the following elements

  return [num[0]] #return the first element


def rest(num):
  return num[len(longest(num))]


def merge(left, right):
  if len(left) == 0:
    return right

  if len(right) == 0:
    return left

  if left[0] <= right[0]:
    return [left[0]] + merge(left[1:], right)

  else:
    return [right[0]] + merge(right[1:], left)


def sort(num):
  if len(num) <= 1:
    return num

  return merge(longest(num), sort(rest(num)))

```

# **Quicksort**

```python
def quicksort(num):
  if len(num) <= 1:
    return num

  pivot = num[0]
  smaller = []
  equal = []
  larger = []

  for number in num:
    if num < pivot:
      smaller.append(num)

    elif num == pivot:
      equal.append(num)

    else:
      larger.append(num)

  return quicksort(smaller) + equal + quicksort(larger)
```

```python
def longest(num):
  if len(num) <= 1:
    return num

  if num[0] <= num[1]:
    return [num[0]] + longest(num[1:])

  return [num[0]]

def rest(num):
  return num[len(longest(num)):]

def merge(left, right):
  if len(right) == 0:
    return left

  if len(left) == 0:
    return right

  if left[0] <= right[0]:
    return [left[0]] + merge(left[1:], right)

  else:
    return [right[0]] + merge(left, right[1:])

def sort(num):
  if len(num) <= 1:
    return num

  return merge(longest(num), sort(rest(num)))

print(sort([2, 5, 7, 2, 4, 1]))
```
