# **3 - Part 2. Implement a LinkedList ADT in python. Include tests and expected outcomes**

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class LinkedList:
  def __init__(self):
    self.head = None

  def insert(self, value):
    new_node = Node(value)
    if self.head is None:
      self.head = new_node
      return
    current = self.head
    while current.next:
      current = current.next
    current.next = new_node

  def find(self, value):
    current = self.head
    while current:
      if current.data == value:
        return True
      current = current.next
    return False

  def remove(self, value):
    if self.head is None:
      return False
    if self.head.data == value:
      self.head = self.head.next
      return True
    current = self.head
    while current.next:
      if current.next.data == value:
        current.next = current.next.next
        return True
      current = current.next
    return False

  def show(self):
    current = self.head
    while current:
      print(current.data, end=" -> ")
      current = current.next
    print("None")
```

**Towers of Hanoi**

```python
def hanoi(n, start, helper, target):

  if n == 1:
    print(start, "->", target) #if theres only 1 disk, just move it from start to target

  else:
    hanoi(n-1, start, helper, target) #take total number of disks minus the bottom one, move them from start, to the helper, using target

    print(start, "->", target)

    hanoi(n-1, helper, target, start) #move the disks from helper, to target, using start
```

**Quicksort**

```python
def quicksort(seq):

  if not seq:
    return []

  pivot = seq[0] #the pivot is the first element in the seq

  left = [x for x in seq[1:] if x <= pivot] #For any number after the pivot (1:), if smaller than pivot, add to left list

  right = [x for x in seq[1:] if x > pivot] #For any number after the pivot, if larger than pivot, add to the right list

  return quicksort(left) + [pivot] + quicksort(right) #return the left list (smaller than pivot) + pivot + right list (larger than pivot), this is the sorted list
```

**Sort using longest + merge**

```python
#longest
def longest(seq):

  if len(seq) <= 1: #if the length is less than 1, return the seq (only 1 number)
    return seq

  if seq[0] <= seq[1]: #if first element is less than the next
    return [seq[0]] + longest(seq[1:]) #return first element and run the longest again for the following elements

  return [seq[0]] #return the first element
```

```python
#rest
def rest(seq):
  return seq[len(longest(seq)):]
```

```python
#merge
def merge(a, b):

  if not a:
    return b #if theres no a, then return b

  if not b:
    return a #if theres no b, then return a

  if a[0] <= b[0]: #first element in a is less than or equal to first element in b
    return [a[0]] + merge(a[1:], b) #return that first element in a and recall the merge function starting at element 1 in a

  return [b[0]] + merge(a, b[1:]) #if element in a is larger than b, return that element in b and recall merge function with next element in b
```

```python
def sort(seq):

  if len(seq) <= 1:
    return seq

  return merge(longest(seq), sort(rest(seq)))
  #longest(seq) should already be sorted due to the longest code having shortest to largest already
  #sort the rest list
  #merge the 2 lists together
```

**Fraction Class**

```python
class Fraction:
  def __init__(self, num, den):
    self.num = num
    self.den = den

  def __str__(self):
    return str(self.num) + "/" + str(self.den)

  def __add__(self, other): #f1 + f2 calls __add__
    return Fraction(self.num * other.den + self.den * other.num, self.den * other.den)

  def __mul__(self, other): #f1 * f2 calls __mul__
    return Fraction(self.num * other.num, self.den * other.den)
```

# **LINKED LIST**

A linked list works like a treasure map, instead of clues, we have Nodes. Each node stores some data on where the next node is.

[10|next]->[20|next]->[30|next]->None

10 knows where 20 is. 20 knows where 30 is. 30 knows there is nothing after it.

**Node**

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None
```

Node = box

Each box contains a value and a pointer to the next box

n = Node(10) is equal to

n.data = 10

n.next = None

**Insert**

```python
def insert(self, value): #insert(10)

  new = Node(value) #make new node [10]

  if self.head is None: #if head is empty
    self.head = new #head is new node -> head = [10]
    return

  current = self.head #current is the head [10]

  while current.next: #walk to the end (with only one node, 10.next = none)
    current = current.next #current is the next node

  current.next = new #attach node 10 -> 20
```

Goal is to add something to the list

*First insert:*

insert(10) -> [10] -> since there is no head yet ->

if self.head is None:

self.head = new

So now: head -> [10]

-------------------------------

*Second Insert:*

insert(20) -> current list is 10 but we need 10 -> 20

**current = self.head**

means current = 10

**while current.next:**

means can I keep moving? Currently 10.next = None

**current.next = new**

means 10 -> 20

**Find**

```python
def find(self, value):

  current = self.head #start at the head

  while current:

    if current.data == value: #if current = value, return true
      return True

    current = current.next #if not, check next value and so on

  return False #if current never = value, return false
```

Goal is to find if a value exists

10 -> 20 -> 30

find(20)

-------------------------------

Start: current = self.head

current -> 10 -> 20 -> 30

Check: if current.data == value

10 == 20? No.

Move: current = current.next

20 == 20? Yes.

Return: True

Return false if value does not exist

**Show**

```python
def show(self):

  current = self.head #sets current to the head (10 in this case)

  while current:
    print(current.data) #print current data -> 10
    current = current.next #set current to the next 10 -> 20
```

Goal is to print everything

current = self.head -> 10

print(current.data) -> 10

**Remove**

```python
def remove(self, value):

  if self.head.data == value: #if head is equal to the removed value remove(10)
    self.head = self.head.next #head becomes the next value head -> 20
    return

  current = self.head #current value is the head current -> 20

  while current.next:

    if current.next.data == value: #if current is equal to removed value remove(10)
      current.next = current.next.next #current.next skips over removed value and becomes current.next.next
      return

    current = current.next
```

Goal is not to remove a node but to skip over a value
