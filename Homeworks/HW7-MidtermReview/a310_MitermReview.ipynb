**Define suprervised machine learning:**

- Supervised machine learning is a type of learning used to make predictions from existing patterns in labeled data. It is used for regression and classification problems

**Define unsupervised machine learning:**

- Unsupervised machine learning is used to discover hidden patterns, structures, or relationships in data without known labels or target outputs

**What is linear regression:**

- Linear regression is a supervised learning method that finds the best fit line relating predictor variables to an outcome variable and uses that line to make predictions

**What is logistic regression:**

- Logistic regression is a supervised classification method that predicts the probability that an observation belongs to a particular category, often a binary outcome such as yes/no or true/false

**What is k-means clustering:**

- K-means clustering is an unsupervised learnign algorithm that groups similar data points into *k* clusters by repeatedly assigning points to the nearest cluster and updating those centers

**What is k-nearest neighbors:**

- K-nearest neighbors is a supervised learning algorithm that classifies or predicts a data point based on the values or classes of its *k* closest neighbors

**What is an artificial neural network:**

- An artificial neural network is a machine learning model made of interconnected layers of artifical neurons that learn patterns from data and make predictions through weighted connections and training

**Define basic statistics of average, variance, and standard deviation:**

- Average: Sum of all values divided by the number of values
- Variance: Measures how spread out data values are from the mean
- Standard deviation: The square root of the variance; measures typical distance from the mean in the same units of data

**What is a support vector machine:**

- A support vector machine is a supervised learning algorithm that separates data into classes by finding the optimal boundary between groups using support vectors

# **3 - Part 1. Write a program that would determine experimentally the probability of getting six heads in a row in 100 coin flips.** **bold text**

```python
import random

TRIALS = 100000
successes = 0

for _ in range (TRIALS) :
  flips = "".join(random.choice("HT") for _ in range(100))

  if "HHHHHH" in flips:
    successes += 1

print(successes / TRIALS)
```

The probability of getting six consectutive heads somewhere within 100 flips is somewhat high with a roughly 54% chance of happening.

# **3 - Part 2. Implement a LinkedList ADT in python. Include tests and expected outcomes**

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class LinkedList:
  def __init__(self):
    self.head = None

  def insert(self, value):
    new_node = Node(value)
    if self.head is None:
      self.head = new_node
      return
    current = self.head
    while current.next:
      current = current.next
    current.next = new_node

  def find(self, value):
    current = self.head
    while current:
      if current.data == value:
        return True
      current = current.next
    return False

  def remove(self, value):
    if self.head is None:
      return False
    if self.head.data == value:
      self.head = self.head.next
      return True
    current = self.head
    while current.next:
      if current.next.data == value:
        current.next = current.next.next
        return True
      current = current.next
    return False

  def show(self):
    current = self.head
    while current:
      print(current.data, end=" -> ")
      current = current.next
    print("None")
```

**Tests:**

```python
ll = LinkedList()
ll.insert(10)
ll.insert(20)
ll.insert(30)

ll.show()
print(ll.find(20))
print(ll.find(50))

ll.remove(20)

ll.show()
print(ll.find(20))
```

# **5. define a longest sorted prefix (ascending) function that works on a list of numbers. Then define a recursive sort method that uses it (along with some other helpers):**

**- longest(seq) examples:
Returns the longest ascending prefix at the front of the list.**

```python
longest([4, 5, 2, 3, 1, 7, 6])
# [4, 5] because the ascending order is 4-->5, then cuts off at 2 since its smaller than 5

longest([1, 2, 3, 4])
# [1, 2, 3, 4] because 1-->4 is already ascending order

longest([5, 3, 4])
# [5] because 3 is lower than 5 so it stops after 5
```

Python code version

```python
def longest(seq):
    if len(seq) <= 1:
        return seq

    if seq[0] <= seq[1]:
        return [seq[0]] + longest(seq[1:])
    else:
        return [seq[0]]
```

**- rest(seq) examples:
Returns everything left after removing the longest sorted prefix.**

```python
rest([4, 5, 2, 3, 1, 7, 6])
# [2, 3, 1, 7, 6] basically the opposite of longest(seq), it prints what is left after removing 4 and 5

rest([1, 2, 3, 4])
# [] because 1, 2, 3, and 4 are in order and there is nothing after

rest([5, 3, 4])
# [3, 4] because since 5 is the only ascending order, 3 and 4 is whats left
```

Python code version

```python
def rest(seq):
  return seq[len(longest(seq)):]
```

**- merge(a, b): Standard merge from merge sort**

Python code version

```python
def merge(a, b):
  if not a:
    return b

  if not b:
    return a

  if a[0] <= b[0]:
    return [a[0]] + merge(a[1:], b)

  else:
    return [b[0]] + merge(a, b[1:])
```
