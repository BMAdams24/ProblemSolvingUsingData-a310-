Stage **1**

```python
class Given:

    def __init__(self, key):
        self.key = key
        self.right = None
        self.left = None

    def display(self):
        lines, *_ = self._display_aux()
        for line in lines:
            print(line)

    def _display_aux(self):

        if self.right is None and self.left is None:
            line = '%s' % self.key
            width = len(line)
            height = 1
            middle = width // 2
            return [line], width, height, middle

        if self.right is None:
            lines, n, p, x = self.left._display_aux()
            s = '%s' % self.key
            u = len(s)
            first_line = (x + 1) * ' ' + (n - x - 1) * '_' + s
            second_line = x * ' ' + '/' + (n - x - 1 + u) * ' '
            shifted_lines = [line + u * ' ' for line in lines]
            return [first_line, second_line] + shifted_lines, n + u, p + 2, n + u // 2

        if self.left is None:
            lines, n, p, x = self.right._display_aux()
            s = '%s' % self.key
            u = len(s)
            first_line = s + x * '_' + (n - x) * ' '
            second_line = (u + x) * ' ' + '\\' + (n - x - 1) * ' '
            shifted_lines = [u * ' ' + line for line in lines]
            return [first_line, second_line] + shifted_lines, n + u, p + 2, u // 2

        left, n, p, x = self.left._display_aux()
        right, m, q, y = self.right._display_aux()

        s = '%s' % self.key
        u = len(s)

        first_line = (
            (x + 1) * ' ' +
            (n - x - 1) * '_' +
            s +
            y * '_' +
            (m - y) * ' '
        )

        second_line = (
            x * ' ' +
            '/' +
            (n - x - 1 + u + y) * ' ' +
            '\\' +
            (m - y - 1) * ' '
        )

        if p < q:
            left += [n * ' '] * (q - p)
        elif q < p:
            right += [m * ' '] * (p - q)

        zipped_lines = zip(left, right)

        lines = [
            first_line,
            second_line
        ] + [a + u * ' ' + b for a, b in zipped_lines]

        return lines, n + m + u, max(p, q) + 2, n + u // 2


b = Given(1)

print("-------------", 1)
b.display()
print("end of first test")

c = Given(3)

print("-------------", 3)
c.display()

c.left = Given(4)

print("-------------", 4)
c.display()

c.right = Given(5)

print("-------------", 5)
c.display()

print("end of second test")
```

**Stage 5**

```python
class Heap(Given):

    def __init__(self, key):
        super().__init__(key)

    def size(self):

        if self.left is None and self.right is None:
            return 1

        elif self.left is None:
            return 1 + self.right.size()

        elif self.right is None:
            return 1 + self.left.size()

        else:
            return 1 + self.left.size() + self.right.size()

    def insert(self, value):

        path = format(self.size() + 1, "b")

        self.helper(path[1:], value)

        self.clean()

    def helper(self, path, value):

        if path == "0":
            self.left = Heap(value)

        elif path == "1":
            self.right = Heap(value)

        else:

            if path[0] == "0":
                self.left.helper(path[1:], value)
            else:
                self.right.helper(path[1:], value)

    def clean(self):

        if self.left is not None:
            self.left.clean()

        if self.right is not None:
            self.right.clean()

        if self.left is None and self.right is None:
            pass

        elif self.right is None:

            if self.key > self.left.key:
                self.key, self.left.key = self.left.key, self.key

        else:

            if self.key <= self.left.key and self.key <= self.right.key:
                pass

            elif self.left.key < self.right.key:
                self.key, self.left.key = self.left.key, self.key

            else:
                self.key, self.right.key = self.right.key, self.key

    def helperRemove(self, path):

        if path == "":
            return self.key

        elif path == "0":
            return self.left.key

        elif path == "1":
            return self.right.key

        else:

            if path[0] == "0":
                return self.left.helperRemove(path[1:])
            else:
                return self.right.helperRemove(path[1:])

    def cleanRemove(self):

        if self.left is None and self.right is None:
            return

        elif self.right is None:

            if self.key > self.left.key:
                self.key, self.left.key = self.left.key, self.key

        else:

            if self.key <= self.left.key and self.key <= self.right.key:
                return

            if self.left.key < self.right.key:

                self.key, self.left.key = self.left.key, self.key

                self.left.cleanRemove()

            else:

                self.key, self.right.key = self.right.key, self.key

                self.right.cleanRemove()

    def removeTop(self):

        print("Removing the top:", self.key)

        path = format(self.size(), "b")

        value = self.helperRemove(path[1:])

        self.key = value

        p = self

        for c in path[1:-1]:

            if c == "0":
                p = p.left
            else:
                p = p.right

        if path[-1] == "0":
            p.left = None
        else:
            p.right = None

        self.cleanRemove()

        return self


print("We start with insertions, from the empty heap...\n")

a = Heap(100)

a.display()

for i in range(99, 92, -1):

    print("------------- insert", i)

    a.insert(i)

    a.display()

print("\nNow we start removing the top value...")
print("... notice how the temporary top percolates down.\n")

for i in range(5):

    a = a.removeTop()

    a.display()
```

**Stage 6**

```python
def removeTop(heap):

    print("Remove the top value in the heap.")

    if heap is None:
        return None

    elif heap.left is None and heap.right is None:
        return None

    else:
        return heap.removeTop()


def display(heap):

    print("The heap becomes:")

    if heap is None:
        print(None)

    else:
        heap.display()


print("We start with insertions, from the empty heap...\n")

a = Heap(100)

a.display()

for i in range(99, 92, -1):

    print("------------- insert", i)

    a.insert(i)

    a.display()

print("\nNow we start removing the top value...")
print("... notice how the temporary top percolates down.\n")

size = a.size()

for i in range(size + 3):

    a = removeTop(a)

    display(a)

    print("----------------------------")
```

```python
import random


def removeTop(heap):

    if heap is None:
        return None

    elif heap.left is None and heap.right is None:
        return None

    else:
        return heap.removeTop()


def display(heap):

    if heap is None:
        print(None)

    else:
        heap.display()


b = Heap(random.randrange(-50, 50))

b.display()

for _ in range(8):

    value = random.randrange(-50, 50)

    print("----------------( now inserting", value, ")--")

    b.insert(value)

    b.display()


for _ in range(12):

    print("Removing top of heap now ... heap becomes:")

    b = removeTop(b)

    display(b)
```
